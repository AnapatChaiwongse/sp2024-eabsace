### Import File

```python
import time
from difflib import SequenceMatcher
from BackTranslation import BackTranslation
from requests import ReadTimeout
from langdetect import detect
```

```python
# ! Check File Name !
FOLD = "1"
INPUT_PATH = f"dataset/train1/splited-{FOLD}.txt"
OUTPUT_PATH_DE = f"/output/fold-{FOLD}/eng_german.txt"
OUTPUT_PATH_FR = f"/output/fold-{FOLD}/eng_french.txt"
OUTPUT_PATH_JA = f"/output/fold-{FOLD}/eng_japanese.txt"
OUTPUT_PATH_ES = f"/output/fold-{FOLD}/eng_spanish.txt"

f = open(INPUT_PATH, encoding="utf8")
text = f.readlines() 
```

### Split Data

```python
sentences = []
sentiments = []

for x in text:
    splited = x.split('\x01')
    sentences.append(splited[0])
    sentiments.append(splited[1])
```

```python
print("sentences Sentences: " + str(len(sentences)))
print("Sentiments Sentences: " + str(len(sentiments)))
```

### Detect language

```python
english_sentences = []
thai_sentences = []
english_sentiment = []
thai_sentiment = []
count = 0

for sentence in sentences:
    try:
        language = detect(sentence)
        if language == 'th':
            thai_sentences.append(sentence)
            thai_sentiment.append(sentiments[count])
        else:
            english_sentences.append(sentence)
            english_sentiment.append(sentiments[count])
    except Exception as e:
        print(f"Error: {e}\nCould not detect language for sentence: {sentence}") 
    count = count+1
```

```python
print("Eng Sentences: " + str(len(english_sentences)))
print("Thai Sentences: " + str(len(thai_sentences)))
print("Sum: " + str(len(english_sentences) + len(thai_sentences)))
```

### Back Translation

#### Eng - German

```python
trans = BackTranslation()

result = []
backtranslated_sentence = []
max_retries = 3

for i in range(len(english_sentences)):
    
    retries = 0
    while retries < max_retries:
        try:
            result = trans.translate(english_sentences[i], src='en', tmp='de', sleeping=1)
            backtranslated_sentence.append(result.result_text)
            break  
        except ReadTimeout:
            retries += 1
            print(f"ReadTimeout error. Retrying ({retries}/{max_retries})...")
            time.sleep(2)  
    else:
        print(f"Failed to translate sentence after {max_retries} retries: {english_sentences[i]}")
```

```python
eng_german = []

with open(OUTPUT_PATH_DE, "w", encoding="utf-8") as file:
    for item1,item2 in zip(backtranslated_sentence, english_sentiment):
        file.write(item1 + "" + item2 )
        eng_german.append(f"{item1}{item2}")
```

```python
# Calculate Change Percentage and Print Results of Original and Backtranslated Sentences
total_similarity = 0  
total_change = 0 

for idx in range(len(english_sentences)):
    similarity = SequenceMatcher(None, english_sentences[idx], backtranslated_sentence[idx]).ratio() * 100
    change_percentage = 100 - similarity
    
    total_similarity += similarity
    total_change += change_percentage
    
    print(f"Original sentence: {english_sentences[idx]}")
    print(f"Backtranslated: {backtranslated_sentence[idx]}")
    print(f"Similarity: {similarity:.2f}%")
    print(f"Change Percentage: {change_percentage:.2f}%")
    print("-------------------------------------------------------------")

average_similarity = total_similarity / len(english_sentences)
average_change = total_change / len(english_sentences)

change_ge = average_change

print(f"Average Similarity: {average_similarity:.2f}%")
print(f"Average Change Percentage: {average_change:.2f}%")
```
