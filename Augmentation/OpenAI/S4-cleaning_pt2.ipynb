```python
import re
from difflib import SequenceMatcher
```

## Duplicate Check

```python
texts = []
sentiments = []
ai_texts =[]
ai_sentiments = []

FOLD = 3
OG_PATH = f"../Datasets-{FOLD}-fold/Original/splited-{FOLD}.txt"
PATH = f"outputs/splited-{FOLD}/gpt4o-mini-splited-{FOLD}.txt"

with open(OG_PATH, "r", encoding='utf-8') as file:
    for line in file:
        splited = line.split('')
        texts.append(splited[0])
        sentiments.append(splited[1])

with open(PATH, "r", encoding='utf-8') as file:
    for line in file:
        splited = line.split('')
        ai_texts.append(splited[0])
        ai_sentiments.append(splited[1])
```

```python
filtered_text = []
filtered_sentiments = []

duplicate_count = 0

for i, text in enumerate(ai_texts):
    if text not in texts:
        filtered_text.append(text)
        filtered_sentiments.append(ai_sentiments[i])
    else:
        duplicate_count += 1

with open(PATH, 'w', encoding='utf-8') as file:
    for text, sentiment in zip(filtered_text, filtered_sentiments):
        file.write(f"{text}{sentiment}")
```

```python
print(f"Initial number of gpt4o-mini: {len(ai_texts)}")
print(duplicate_count)
print(f"Afer delete duplicate : {len(filtered_text)}")
```

## Changing Score

```python
texts = []
sentiments = []
ai_texts =[]
ai_sentiments = []

with open(OG_PATH, "r", encoding='utf-8') as file:
    for line in file:
        splited = line.split('')
        texts.append(splited[0])
        sentiments.append(splited[1])

with open(PATH, "r", encoding='utf-8') as file:
    for line in file:
        splited = line.split('')
        ai_texts.append(splited[0])
        ai_sentiments.append(splited[1])
```

```python
count = 1
changeScores = []

assert len(ai_texts) % 3 == 0, "Mismatch: list2 should have 3 sentences per original."

for i, original in enumerate(texts):
    original_clean = re.sub(r'\s+', ' ', original.strip())
    
    generated_group = ai_texts[i*3:(i+1)*3]
    
    for j, generated in enumerate(generated_group):
        generated_clean = re.sub(r'\s+', ' ', generated.strip())
        
        changeScore = round(1 - SequenceMatcher(a=original_clean, b=generated_clean).ratio(), 2) * 100
        changeScores.append(changeScore)

        output = (f"{count}\n"
                  f"Original: {original_clean}\n"
                  f"GPT4o-mini Augmented: {generated_clean}\n"
                  f"Percent of change: {changeScore:.2f}%\n"
                  f"{'-' * 100}\n")
        print(output)
        count+=1
```

```python
if changeScores:
    average_change = round(sum(changeScores) / len(changeScores), 2)
    print(f"Average Percent of Change: {average_change}%")
```
